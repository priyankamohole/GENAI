```python
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model='gemini-3.8-flash')
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
text = "hello my  name is priyanka mohole. Im a data science student.my email is priyankamohole@gmail.com. i am 22 years old."
try:
    res = llm.invoke("give me only name , email and age from the text: " + text)
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
res.content
```

Structured Output

```python
from pydantic import BaseModel,  Field
from typing import List

class ResponseStructure(BaseModel):
    name: str = Field(description="Name from the text")
    email: str = Field(description="Email from the text")
    age: int = Field(description="Age from the text")

structured_llm = llm.with_structured_output(ResponseStructure)
```

```python
res=structured_llm.invoke("give me only name , email and age from the text: " + text)
```

```python
res.age
```

```python
res.model_dump()   # convert to dictionary
```

```python
class Movies(BaseModel):
    title :str = Field(description="Title of the movie")
    year :int =Field(description="Movie release year")


class AllMovies(BaseModel):
    movies: List[Movies]

movie_llm=llm.with_structured_output(AllMovies)
```

```python
res = movie_llm.invoke( "Give me 5 thriller movies with titles and release years.")
```
