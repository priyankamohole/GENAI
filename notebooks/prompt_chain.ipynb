##  Prompt templates - Dynamic prompts

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")
content = StrOutputParser()
```

```python
def transform_case(result:str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "you are a translator that translates in {language}"},
    {"role": "user", "content": "{query}"},
])
```

```python
prompts
```

```python
prmt= prompts.format_messages(language="hindi", query="l love to play football")
prmt
```

```python
res = llm.invoke(prmt)
res.content
```
