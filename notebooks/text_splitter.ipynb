Text Splitting

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text = "Today, I am going to explain my project on a Retrieval-Augmented Generation, also known as RAG, pipeline.RAG is a technique that combines information retrieval with a Large Language Model to generate accurate and relevant answers based on provided documents.In my project, "
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap =20)
```

```python
chunk= splitter.split_text(text)
```

Document Text Loader

```python
from langchain_community.document_loaders import PyPDFLoader , TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/speach.txt")
texts = loader.load()
```

```python
pdfloader = PyPDFLoader("../data/medical_report.pdf")
pdfdata = pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap=100)
# chunks size and chunks_overlap affect on chunks length if i put chunk_size 1000 and overlap 200 chunk len become 26
```

```python
text_chunks = splitter.split_documents(texts)
len(text_chunks)
```

```python
pdf_chunks = splitter.split_documents(pdfdata)
len(pdf_chunks)
```
