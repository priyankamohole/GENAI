Text Splitting

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text = "Today, I am going to explain my project on a Retrieval-Augmented Generation, also known as RAG, pipeline.RAG is a technique that combines information retrieval with a Large Language Model to generate accurate and relevant answers based on provided documents.In my project, "
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap =20)
```

```python
chunk= splitter.split_text(text)
```
