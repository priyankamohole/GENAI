```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import OpenAIEmbeddings
from langchain_ollama import OllamaEmbeddings
```

```python
document = ["Today, I am going to explain my project on a Retrieval-Augmented Generation, "
"also known as RAG, pipeline.RAG is a technique that combines information retrieval with a Large Language Model "
"to generate accurate and relevant answers based on provided documents.In my project"]
```

```python
embedding = OllamaEmbeddings(model="all-minilm")
```

```python
vector = embedding.embed_documents(document)
```

```python
len(vector[0])
# vector
```

Store Vector in Chroma DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=document,
    embedding=embedding,
    persist_directory="./vector_db"
)
```

```python
query = "RAG"
result = vector_store.similarity_search(query=query, k = 2)  
# k is how many result to get , response count
result[0].page_content
```
