```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """
    it ewill return the sum of two numbers a and b
    Args : 
        a: number 1
        b: number 2
    """
    return a + b

@tool
def multiply_numbers(a:int, b:int):
    """
    it will return the product of two numbers a and b
    Args : 
        a: number 1
        b: number 2
    """
    return a * b
```

```python
add_numbers.invoke({"a":5, "b":10})
# add_numbers.invoke(4, 5)  error
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash" )
```

```python
agent = create_agent(
    model = llm , 
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a math teacher and always use tools for calculations"
)
```

```python
response = agent.invoke({"messages":[{"role":"user", "content":"what is the sum of 5 and 14"}]})
print(response['messages'][-1].text)
```
